## **Planning**

## **Title of Proposed Final Project: Hybrid Book Recommendation System**

For this final project, I plan to develop a hybrid book recommendation system using the Book Recommendation Dataset from Kaggle: https://www.kaggle.com/datasets/arashnic/book-recommendation-dataset

The recommendation system will combine collaborative filtering and content-based filtering to improve the quality of recommendations being provided to avid book worms. The system will be implemented using Python and Spark.

## **Problem Statement**

The purpose of this project is to analyze whether combining collaborative filtering and content-based filtering can potentially improve recommendation quality compared to traditional recommender system methods that depend on user-item interactions to identify preferences in user preferences.

## **Dataset**

The Book Recommendation Dataset from Kaggle will be utilized to build the recommender system.

The dataset contains:

- Approximately 278,858 users

- Approximately 271,379 books

- Approximately 1,149,780 ratings

The dataset includes three main files:

1) **Ratings Dataset**

Contains user-book interactions:

- User ID
- ISBN
- Rating

This dataset will be used for collaborative filtering.


2) **Books Dataset**

Contains book information:

- ISBN
- Book title
- Author
- Publisher
- Year of publication
- Cover image URL

This dataset will be used for content-based recommendation.

3) **Users Dataset**

Contains user information:

- User ID
- Age
- Location

This information might be used for exploratory analysis.

## **Project Objectives**

The goals of this project are:

- Process a large-scale recommendation dataset using Apache Spark.
- Build a collaborative filtering recommendation model.
- Build a content-based recommendation model.
- Develop a hybrid recommendation approach.
- Compare recommendation performance across different models.
- Evaluate the effectiveness of combining user behavior and item characteristics.

## **Model Evaluation**

The recommendation models will be evaluated using:

- Root Mean Squared Error (RMSE)

  Measure the difference between predicted and actual ratings.

- Precision@K

  Measure how many recommended books are relevant.

- Recall@K

  Measure how many relevant books are successfully recommended.
